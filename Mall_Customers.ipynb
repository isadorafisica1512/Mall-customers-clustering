```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Woman between 18 and 30

```python
dataset = pd.read_csv('Mall_Customers.csv')
dataset = dataset[dataset['Genre'] != 'Male']
dataset = dataset.drop('CustomerID', axis=1)
dataset = dataset[(dataset['Age'] >= 18) & (dataset['Age'] <= 30)]

print(dataset.head())
```

```python
x = dataset.iloc[:, [2,3]]
```

```python
print(x)
```

```python
from sklearn.cluster import KMeans
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, init = 'k-means++', random_state = 42)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
kmeans = KMeans(n_clusters = 5, init = 'k-means++', random_state = 42)
y_kmeans = kmeans.fit_predict(x)
print(y_kmeans)
```

```python
plt.scatter(x.iloc[y_kmeans == 0, 0],x.iloc[y_kmeans == 0, 1])
plt.scatter(x.iloc[y_kmeans == 1, 0],x.iloc[y_kmeans == 1, 1])
plt.scatter(x.iloc[y_kmeans == 2, 0],x.iloc[y_kmeans == 2, 1])
plt.scatter(x.iloc[y_kmeans == 3, 0],x.iloc[y_kmeans == 3, 1])
plt.scatter(x.iloc[y_kmeans == 4, 0],x.iloc[y_kmeans == 4, 1])
plt.title('Woman between 18 and 30')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')

```

**Identified Customer Clusters**

- **Low Income / High Spending:** Customers with relatively low annual income but high spending scores.
- **Moderate Income / Moderate Spending:** Customers with intermediate income and spending levels.
- **Moderate-to-High Income / Lower Spending:** Customers with relatively higher income but moderate or low spending scores.
- **High Income / High Spending:** Customers with high annual income and high spending scores.

**Key Insights**

This group presents diverse consumption patterns. Some customers demonstrate high spending behavior despite having relatively low incomes, while others have greater purchasing power but comparatively lower spending scores.

The presence of both high-income, high-spending and high-income, lower-spending customers suggests that purchasing power alone does not explain consumer behavior.

**Business Recommendations**

### 7. Overall Business Insights

By comparing the six demographic groups, several broader patterns emerge regarding the relationship between annual income and consumer spending behavior. These findings provide a more comprehensive understanding of how customer profiles vary across age groups and gender, highlighting opportunities for more specific marketing strategies.

**1. Higher Income Does Not Always Translate into Higher Spending**

One of the most consistent findings across all demographic groups is that higher annual income does not necessarily correspond to higher spending scores.

In every demographic category, there are customers with relatively high incomes who demonstrate moderate or low spending behavior. This suggests that purchasing power and actual consumption are two different aspects of customer behavior.

For businesses, this distinction is particularly important. Customers with higher incomes should not automatically be classified as high-value customers based on their purchasing power alone. Their actual engagement and spending patterns should also be considered when developing marketing strategies.

**2. Women and Men Present Different Spending Patterns Across Age Groups**

The comparison between genders reveals interesting differences in customer segmentation.

Among customers aged 18–30, both men and women present diverse spending profiles, including customers with relatively low incomes but high spending scores, as well as higher-income customers with lower spending behavior.

Among customers aged 30–45, women exhibit a broader distribution of customer profiles, ranging from low-income, low-spending customers to high-income, high-spending customers. Men in the same age group appear more concentrated in three main segments, particularly moderate-to-high-income customers with high spending scores.

Among customers aged 45 and above, an interesting difference emerges. Women include a segment of high-income customers with moderate-to-high spending scores, whereas the high-income segment identified among men is characterized by comparatively low spending scores.

This suggests that the relationship between income and spending behavior may vary considerably across gender and age categories.

**3. Women Aged 18–30 Present Diverse Consumption Profiles**

Young women represent an interesting demographic group because their spending behavior varies considerably across income levels.

Some customers demonstrate high spending scores despite having relatively low incomes, while others with higher purchasing power exhibit more moderate or lower spending behavior.

This indicates that young women's consumption patterns cannot be explained by income alone. Businesses may benefit from offering different strategies for this audience, combining accessible promotions and loyalty programs with personalized recommendations and premium experiences.

**4. Customers Aged 30–45 Represent an Important Market Segment**

The 30–45 age group presents particularly interesting opportunities for businesses.

Women in this category demonstrate a wide range of income and spending combinations, suggesting the presence of multiple customer profiles with different purchasing capacities and consumption behaviors.

Men in the same age group show a notable concentration of moderate-to-high-income, high-spending customers. This segment represents an interesting opportunity for customer retention, premium offers, and loyalty programs.

For businesses, these findings suggest that customers within the same age category may still require substantially different marketing approaches depending on their income and spending profiles.

**5. Higher Income Among Women Aged 45+ Is Associated with More Consistent Spending Potential**

An interesting pattern emerges among women aged 45 and above.

Unlike men in the same age category, whose high-income segment is characterized by low spending scores, women include a high-income segment with moderate-to-high spending behavior.

Additionally, women with moderate incomes and high spending scores represent another relevant customer segment.

This suggests that businesses may find valuable opportunities among women aged 45 and above, particularly through premium products, personalized shopping experiences, and customer retention initiatives.

However, the presence of high-income, low-spending women also reinforces the importance of distinguishing purchasing power from actual customer engagement.

**6. Men Aged 45+ Show a Stronger Concentration of Moderate-Income, High-Spending Customers**

Among men aged 45 and above, one of the most noticeable segments consists of customers with moderate incomes and high spending scores.

This indicates that strong customer engagement is not necessarily associated with the highest income levels.

At the same time, the presence of high-income, low-spending customers suggests that businesses targeting this demographic should not rely exclusively on premium pricing or purchasing power as indicators of potential demand.

Marketing strategies focused on product relevance, customer preferences, and personalized engagement may be more effective than simply offering more expensive products.

**7. High-Spending Customers Can Be Found Across Different Income Levels**

Another important finding is the presence of high-spending customers across several income categories and demographic groups.

Customers with relatively low or moderate incomes frequently demonstrate high spending scores, indicating that strong consumption engagement is not exclusive to wealthier customers.

From a business perspective, this means that customer value should not be evaluated exclusively through annual income.

Businesses should also consider spending behavior when identifying valuable customer segments, designing loyalty programs, and allocating marketing resources.

**8. Different Demographic Groups Require Different Marketing Approaches**

The overall analysis suggests that a single marketing strategy may not be equally effective across all customer segments.

For example:

- **Younger customers:** Accessible promotions, loyalty rewards, personalized recommendations, and engaging shopping experiences may help maintain customer engagement.
- **Women aged 30–45:** A diversified marketing approach may be appropriate, considering the wide range of income and spending profiles identified in this group.
- **Men aged 30–45:** Customer retention strategies may be particularly valuable for the moderate-to-high-income, high-spending segment.
- **Women aged 45+:** Premium offers, personalized services, and loyalty initiatives may help businesses engage customers with higher purchasing power and strong spending behavior.
- **Men aged 45+:** Retention strategies for moderate-income, high-spending customers and targeted engagement campaigns for high-income, low-spending customers may represent relevant opportunities.

Ultimately, demographic characteristics can help businesses identify broad customer patterns, but income and spending behavior provide additional information for developing more specific marketing strategies.
